```python
from pathlib import Path

import pandas as pd
from rdflib import Graph
```

```python
PROJECT_ROOT = Path.cwd().parent.parent

ONTOLOGY_PATH = (
    PROJECT_ROOT
    / "owl"
    / "ontology_v23.rdf"
)

print("Thư mục hiện tại:", Path.cwd())
print("Đường dẫn ontology:", ONTOLOGY_PATH.resolve())
print("File tồn tại:", ONTOLOGY_PATH.exists())

if not ONTOLOGY_PATH.exists():
    raise FileNotFoundError(
        f"Không tìm thấy file RDF: {ONTOLOGY_PATH.resolve()}"
    )

graph = Graph()
graph.parse(ONTOLOGY_PATH)

print(f"Đọc ontology thành công: {len(graph):,} bộ ba RDF")
```

```python
NS_URI = (
    "http://www.semanticweb.org/henrydao/"
    "ontologies/2025/7/TrainingProgramOntology#"
)

PREFIXES = f"""
PREFIX : <{NS_URI}>
PREFIX rdf: <http://www.w3.org/1999/02/22-rdf-syntax-ns#>
PREFIX rdfs: <http://www.w3.org/2000/01/rdf-schema#>
PREFIX owl: <http://www.w3.org/2002/07/owl#>
PREFIX xsd: <http://www.w3.org/2001/XMLSchema#>
"""


def run_sparql(query):
    """Chạy truy vấn SPARQL và chuyển kết quả thành DataFrame."""

    if "graph" not in globals():
        raise RuntimeError(
            "Biến graph chưa được khởi tạo. "
            "Hãy chạy cell đọc ontology trước."
        )

    result = graph.query(PREFIXES + query)

    rows = [
        {
            str(variable): (
                str(value)
                if value is not None
                else None
            )
            for variable, value in row.asdict().items()
        }
        for row in result
    ]

    return pd.DataFrame(rows)
```

```python
print("Biến graph tồn tại:", "graph" in globals())
print("Tổng số bộ ba:", len(graph))
```

```python
query_otd_01 = """
SELECT ?component ?quantity
WHERE {
    {
        SELECT
            ("Class" AS ?component)
            (COUNT(DISTINCT ?item) AS ?quantity)
        WHERE {
            ?item rdf:type owl:Class .
        }
    }
    UNION
    {
        SELECT
            ("Object Property" AS ?component)
            (COUNT(DISTINCT ?item) AS ?quantity)
        WHERE {
            ?item rdf:type owl:ObjectProperty .
        }
    }
    UNION
    {
        SELECT
            ("Datatype Property" AS ?component)
            (COUNT(DISTINCT ?item) AS ?quantity)
        WHERE {
            ?item rdf:type owl:DatatypeProperty .
        }
    }
    UNION
    {
        SELECT
            ("Course" AS ?component)
            (COUNT(DISTINCT ?item) AS ?quantity)
        WHERE {
            ?item rdf:type :Course .
        }
    }
    UNION
    {
        SELECT
            ("Student" AS ?component)
            (COUNT(DISTINCT ?item) AS ?quantity)
        WHERE {
            ?item rdf:type :Student .
        }
    }
    UNION
    {
        SELECT
            ("CourseAttempt" AS ?component)
            (COUNT(DISTINCT ?item) AS ?quantity)
        WHERE {
            ?item rdf:type :CourseAttempt .
        }
    }
    UNION
    {
        SELECT
            ("CourseResult" AS ?component)
            (COUNT(DISTINCT ?item) AS ?quantity)
        WHERE {
            ?item rdf:type :CourseResult .
        }
    }
    UNION
    {
        SELECT
            ("StudentTermRecord" AS ?component)
            (COUNT(DISTINCT ?item) AS ?quantity)
        WHERE {
            ?item rdf:type :StudentTermRecord .
        }
    }
}
"""

otd_01 = run_sparql(query_otd_01)
otd_01
```

```python
# kiểm tra quan hệ tiên quyết
query_otd_04_all = """
SELECT
    ?courseCode
    ?courseName
    ?prerequisiteCode
    ?prerequisiteName
WHERE {
    ?course rdf:type :Course ;
            :courseCode ?courseCode ;
            :hasPrerequisiteCourse ?prerequisite .

    OPTIONAL {
        ?course :courseName ?courseName .
    }

    ?prerequisite :courseCode ?prerequisiteCode .

    OPTIONAL {
        ?prerequisite :courseName ?prerequisiteName .
    }
}
ORDER BY ?courseCode ?prerequisiteCode
"""

otd_04_all = run_sparql(query_otd_04_all)

print("Số quan hệ tiên quyết:", len(otd_04_all))
otd_04_all
```

```python
query_otd_04_error = """
SELECT DISTINCT
    ?course
    ?courseCode
WHERE {
    ?course rdf:type :Course .

    OPTIONAL {
        ?course :courseCode ?courseCode .
    }

    ?course :hasPrerequisiteCourse+ ?course .
}
ORDER BY ?courseCode
"""

otd_04_error = run_sparql(query_otd_04_error)

print("Số môn tự tiên quyết hoặc nằm trong chu trình:", len(otd_04_error))
otd_04_error
```

```python
# kiểm tra quan hệ song hành
query_otd_05_all = """
SELECT
    ?courseCode
    ?corequisiteCode
WHERE {
    ?course :corequisiteWith ?corequisite .

    OPTIONAL {
        ?course :courseCode ?courseCode .
    }

    OPTIONAL {
        ?corequisite :courseCode ?corequisiteCode .
    }
}
ORDER BY ?courseCode
"""

otd_05_all = run_sparql(query_otd_05_all)

print("Tổng số quan hệ song hành:", len(otd_05_all))
otd_05_all
```

```python
def get_local_name(value):
    if value is None:
        return None

    text = str(value)

    if "#" in text:
        return text.split("#")[-1]

    return text.rstrip("/").split("/")[-1]
otd_06_display = otd_06_all.copy()

otd_06_display["property"] = (
    otd_06_display["property"]
    .apply(get_local_name)
)

otd_06_display["specialization"] = (
    otd_06_display["specialization"]
    .apply(get_local_name)
)

otd_06_display
```

```python
query_otd_06_error = """
SELECT DISTINCT
    ?courseCode
    ?property
    ?specialization
WHERE {
    VALUES ?property {
        :offeredInSpecialization
        :isRequiredForSpecialization
        :isElectiveForSpecialization
    }

    ?course ?property ?specialization .

    OPTIONAL {
        ?course :courseCode ?courseCode .
    }

    FILTER NOT EXISTS {
        ?specialization rdf:type :Specialization .
    }
}
ORDER BY ?courseCode
"""

otd_06_error = run_sparql(query_otd_06_error)

print("Số liên kết chuyên ngành không hợp lệ:", len(otd_06_error))
otd_06_error
```

```python
#Kiểm tra học kỳ khuyến nghị và kỳ mở môn
query_otd_07 = """
SELECT DISTINCT
    ?courseCode
    ?recommendedSemester
    ?semesterNumber
    ?openSemesterType
    ?errorType
WHERE {
    ?course rdf:type :Course .

    OPTIONAL {
        ?course :courseCode ?courseCode .
    }

    OPTIONAL {
        ?course :recommendedInSemester ?recommendedSemester .

        OPTIONAL {
            ?recommendedSemester :semesterNumber ?semesterNumber .
        }
    }

    OPTIONAL {
        ?course :openSemesterType ?openSemesterType .
    }

    {
        FILTER NOT EXISTS {
            ?course :recommendedInSemester ?semester .
        }

        BIND(
            "Thiếu recommendedInSemester"
            AS ?errorType
        )
    }
    UNION
    {
        ?course :recommendedInSemester ?semester .

        FILTER NOT EXISTS {
            ?semester :semesterNumber ?semesterNumberValue .
        }

        BIND(
            "Academic semester thiếu semesterNumber"
            AS ?errorType
        )
    }
    UNION
    {
        ?course :recommendedInSemester ?semester .
        ?semester :semesterNumber ?semesterNumber .

        FILTER (
            xsd:integer(?semesterNumber) < 1
            ||
            xsd:integer(?semesterNumber) > 8
        )

        BIND(
            "semesterNumber ngoài phạm vi 1–8"
            AS ?errorType
        )
    }
    UNION
    {
        FILTER NOT EXISTS {
            ?course :openSemesterType ?openValue .
        }

        BIND(
            "Thiếu openSemesterType"
            AS ?errorType
        )
    }
    UNION
    {
        ?course :openSemesterType ?openSemesterType .

        FILTER (
            xsd:integer(?openSemesterType) NOT IN (1, 2, 12)
        )

        BIND(
            "openSemesterType không hợp lệ"
            AS ?errorType
        )
    }
}
ORDER BY ?courseCode ?errorType
"""

otd_07 = run_sparql(query_otd_07)

print("Số dòng dữ liệu học kỳ cần kiểm tra:", len(otd_07))
otd_07
```

```python
#Đếm số học phần đầy đủ
query_otd_07_count = """
SELECT
    (COUNT(DISTINCT ?course) AS ?courseCount)
    (COUNT(DISTINCT ?courseWithRecommended) AS ?recommendedCount)
    (COUNT(DISTINCT ?courseWithOpenType) AS ?openTypeCount)
WHERE {
    ?course rdf:type :Course .

    OPTIONAL {
        ?courseWithRecommended rdf:type :Course ;
                               :recommendedInSemester ?semester .
    }

    OPTIONAL {
        ?courseWithOpenType rdf:type :Course ;
                            :openSemesterType ?openType .
    }
}
"""

otd_07_count = run_sparql(query_otd_07_count)
otd_07_count
```

```python

#Kiểm tra tính toàn vẹn sinh viên. Kiểm tra cardinality
query_otd_08_cardinality = """
SELECT
    ?student
    (COUNT(DISTINCT ?studentID) AS ?studentIDCount)
    (COUNT(DISTINCT ?major) AS ?majorCount)
    (COUNT(DISTINCT ?academicClass) AS ?classCount)
    (COUNT(DISTINCT ?cohort) AS ?cohortCount)
WHERE {
    ?student rdf:type :Student .

    OPTIONAL {
        ?student :studentID ?studentID .
    }

    OPTIONAL {
        ?student :hasMajor ?major .
    }

    OPTIONAL {
        ?student :belongsToAcademicClass ?academicClass .
    }

    OPTIONAL {
        ?student :belongsToCohort ?cohort .
    }
}
GROUP BY ?student
HAVING (
    COUNT(DISTINCT ?studentID) != 1
    ||
    COUNT(DISTINCT ?major) != 1
    ||
    COUNT(DISTINCT ?academicClass) != 1
    ||
    COUNT(DISTINCT ?cohort) != 1
)
"""

otd_08_cardinality = run_sparql(query_otd_08_cardinality)

print("Sinh viên sai liên kết bắt buộc:", len(otd_08_cardinality))
otd_08_cardinality
```

```python
#Kiểm tra tính toàn vẹn CourseAttempt
query_otd_09_cardinality = """
SELECT
    ?attempt
    (COUNT(DISTINCT ?course) AS ?courseCount)
    (COUNT(DISTINCT ?status) AS ?statusCount)
    (COUNT(DISTINCT ?record) AS ?recordCount)
    (COUNT(DISTINCT ?attemptNumber) AS ?attemptNumberCount)
WHERE {
    ?attempt rdf:type :CourseAttempt .

    OPTIONAL {
        ?attempt :attemptForCourse ?course .
    }

    OPTIONAL {
        ?attempt :hasCourseStatus ?status .
    }

    OPTIONAL {
        ?attempt :attemptInRecord ?record .
    }

    OPTIONAL {
        ?attempt :attemptNumber ?attemptNumber .
    }
}
GROUP BY ?attempt
HAVING (
    COUNT(DISTINCT ?course) != 1
    ||
    COUNT(DISTINCT ?status) != 1
    ||
    COUNT(DISTINCT ?record) != 1
    ||
    COUNT(DISTINCT ?attemptNumber) != 1
)
"""

otd_09_cardinality = run_sparql(query_otd_09_cardinality)

print("CourseAttempt sai liên kết:", len(otd_09_cardinality))
otd_09_cardinality.head(20)
```

```python
#Kiểm tra điểm và attemptNumber
query_otd_09_values = """
SELECT
    ?attempt
    ?grade
    ?attemptNumber
    ?errorType
WHERE {
    ?attempt rdf:type :CourseAttempt .

    {
        ?attempt :grade ?grade .

        FILTER (
            xsd:decimal(?grade) < 0
            ||
            xsd:decimal(?grade) > 10
        )

        BIND(
            "Điểm ngoài phạm vi 0–10"
            AS ?errorType
        )
    }
    UNION
    {
        ?attempt :attemptNumber ?attemptNumber .

        FILTER (
            xsd:integer(?attemptNumber) < 1
        )

        BIND(
            "attemptNumber nhỏ hơn 1"
            AS ?errorType
        )
    }
}
"""

otd_09_values = run_sparql(query_otd_09_values)

print("CourseAttempt có giá trị sai:", len(otd_09_values))
otd_09_values.head(20)
```

```python
#Kiểm tra CourseResult và StudentTermRecord
query_otd_10_record = """
SELECT
    ?record
    (COUNT(DISTINCT ?student) AS ?studentCount)
    (COUNT(DISTINCT ?term) AS ?termCount)
WHERE {
    ?record rdf:type :StudentTermRecord .

    OPTIONAL {
        ?record :recordOfStudent ?student .
    }

    OPTIONAL {
        ?record :recordInTerm ?term .
    }
}
GROUP BY ?record
HAVING (
    COUNT(DISTINCT ?student) != 1
    ||
    COUNT(DISTINCT ?term) != 1
)
"""

otd_10_record = run_sparql(query_otd_10_record)

print("StudentTermRecord vi phạm:", len(otd_10_record))
otd_10_record.head(20)
```

```python
#CourseResult thiếu môn hoặc trạng thái
query_otd_10_result_cardinality = """
SELECT
    ?result
    (COUNT(DISTINCT ?course) AS ?courseCount)
    (COUNT(DISTINCT ?status) AS ?statusCount)
WHERE {
    ?result rdf:type :CourseResult .

    OPTIONAL {
        ?result :forCourse ?course .
    }

    OPTIONAL {
        ?result :hasFinalCourseStatus ?status .
    }
}
GROUP BY ?result
HAVING (
    COUNT(DISTINCT ?course) != 1
    ||
    COUNT(DISTINCT ?status) != 1
)
"""

otd_10_result_cardinality = run_sparql(
    query_otd_10_result_cardinality
)

print(
    "CourseResult sai học phần hoặc trạng thái:",
    len(otd_10_result_cardinality)
)

otd_10_result_cardinality.head(20)
```
